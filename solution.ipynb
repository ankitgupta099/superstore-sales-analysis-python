```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df.dtypes
```

```python
#Total Sales
total_sales = df["Sales"].sum()
print(total_sales)
```

```python
#Total Profit
total_profit = df["Profit"].sum()
print(total_profit)
```

```python
#Total Quantity
total_quantity = df["Quantity"].sum()
print(total_quantity)
```

```python
#Average Sales
average_sales = df["Sales"].mean()
print(average_sales)
```

```python
#Average Profit
average_profit = df["Profit"].mean()
print(average_profit)
```

```python
#Is the profit positive or negative?
df["Profit_Status"] = np.where(df["Profit"] >= 0,"Profit","Loss")
df[["Profit", "Profit_Status"]].head()
```

```python
#Which category has the highest sales?
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
print(category_sales)
```

```python
#Which category generates the highest profit?
category_profit = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)
print(category_profit)
```

```python
#Category-wise quantity?
category_quantity = df.groupby("Category")["Quantity"].sum().sort_values(ascending=False)
print(category_quantity)
```

```python
#Sales by Category
ab=category_sales.plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
ab.bar_label(ab.containers[0], fmt="%.2f")
plt.show()
```

## Sub-Category Analysis

```python
#Top 10 sub-categories by sales?
subcategory_sales = (df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False))
print(subcategory_sales.head(10))
```

```python
#Top 10 by profit?
subcategory_profit = (df.groupby("Sub-Category")["Profit"].sum().sort_values(ascending=False))
print(subcategory_profit)
```

```python
#Which sub-category is making a loss?
Sub_Category_profit = df.groupby("Sub-Category")["Profit"].sum()
Sub_Category_profit[Sub_Category_profit < 0]
```

```python
#What are the company’s top 10 products by sales?
top_products = (df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10))
print(top_products)
```

```python
plt.figure(figsize=(10, 6))
ab=top_products.plot(kind="bar")
plt.title("Top 10 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.xticks(rotation=90)
ab.bar_label(ab.containers[0], fmt="%.2f")
plt.show()
```

## Region Analysis

```python
#Which region has the highest sales?
region_sales = df.groupby("Region")["Sales"].sum()
print(region_sales)
```

```python
#Which region generates the highest profit?
region_profit = df.groupby("Region")["Profit"].sum().sort_values(ascending=False)
print(region_profit)
```

```python
ab=region_sales.plot(kind="bar")
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
ab.bar_label(ab.containers[0], fmt="%.2f")
plt.show()
```

## Customer Segment Analysis

```python
#Which customer segment is the most valuable?
segment_sales = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)
print(segment_sales)
```

```python
#Which customer segment generates the highest total profit?
segment_profit = df.groupby("Segment")["Profit"].sum()
print(segment_profit)
```

```python
ab=segment_profit.plot(kind="bar")
plt.title("Profit by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Profit")
ab.bar_label(ab.containers[0], fmt="%.2f")
plt.show()
```

## Monthly Sales Analysis

```python
df["Month"] = df["Order Date"].dt.to_period("M")
```

```python
#Monthly sales:
monthly_sales = (df.groupby("Month")["Sales"].sum())
```

```python
monthly_sales.plot(kind="line")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

## Yearly Analysis

```python
df["Year"] = df["Order Date"].dt.year
```

```python
#Yearly Sale
yearly_sales = df.groupby("Year")["Sales"].sum().sort_values(ascending=False)
print(yearly_sales)
```

```python
#Yearly Profit
yearly_profit = df.groupby("Year")["Profit"].sum().sort_values(ascending=False)
print(yearly_profit)
```

```python
ab=yearly_sales.plot(kind="bar")
plt.title("Yearly Sales")
plt.xlabel("Year")
plt.ylabel("Sales")
ab.bar_label(ab.containers[0], fmt="%.2f")
plt.show()
```

## Discount vs Profit

```python
#Does giving higher discounts lead to lower profits?
discount_profit = (df.groupby("Discount")["Profit"].mean().sort_values(ascending=False))
print(discount_profit)
```

```python
plt.scatter(df["Discount"], df["Profit"])
plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.show()
```

#### The analysis suggests that higher discounts are generally associated with lower or negative profits. Therefore, excessive discounting may reduce profitability.

# Shipping Analysis

```python
df["Shipping_Days"] = (df["Ship Date"] - df["Order Date"]).dt.days
```

```python
#Average shipping
df["Shipping_Days"].mean()
```

```python
#According to ship mode
shipping_analysis = (df.groupby("Ship Mode")["Shipping_Days"].mean().sort_values(ascending=False))
print(shipping_analysis)
```
